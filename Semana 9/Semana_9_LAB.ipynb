```python
from py_compile import main


def representar_grafo():

    grafo = {
        "A": ["B", "C"],
        "B": ["A", "D", "E"],
        "C": ["A", "F"],
        "D": ["B"],
        "E": ["B", "G"],
        "F": ["C", "G"],
        "G": ["E", "F"]
    }

    print("\nLISTA DE ADYACENCIA\n")

    for centro in grafo:
        print(centro, "->", " - ".join(grafo[centro]))

    return grafo


def recorrido_BFS(grafo, inicio):

    visitados = []
    cola = []
    niveles = {}

    cola.append(inicio)
    visitados.append(inicio)
    niveles[inicio] = 0

    recorrido = []

    while len(cola) > 0:

        actual = cola.pop(0)
        recorrido.append(actual)

        for vecino in grafo[actual]:

            if vecino not in visitados:

                visitados.append(vecino)
                cola.append(vecino)
                niveles[vecino] = niveles[actual] + 1

    print("\nRECORRIDO BFS")
    print(" -> ".join(recorrido))

    print("\nNiveles")

    maximo = max(niveles.values())

    for nivel in range(maximo + 1):

        print("Nivel", nivel, end=": ")

        lista = []

        for centro in niveles:

            if niveles[centro] == nivel:
                lista.append(centro)

        print(", ".join(lista))


def recorrido_DFS(grafo, inicio):

    visitados = []
    recorrido = []

    def DFS(actual):

        visitados.append(actual)
        recorrido.append(actual)

        for vecino in grafo[actual]:

            if vecino not in visitados:
                DFS(vecino)

    DFS(inicio)

    print("\nRECORRIDO DFS")
    print(" -> ".join(recorrido))


def seleccionar_punto_inicial(grafo):

    inicio = input("\nIngrese el centro inicial: ").upper()

    if inicio in grafo:
        return inicio

    else:
        print("Ese centro no existe.")
        return None


def buscar_centro_BFS(grafo):

    inicio = input("\nCentro inicial: ").upper()
    destino = input("Centro que desea buscar: ").upper()

    if inicio not in grafo or destino not in grafo:

        print("Algún centro no existe.")
        return

    cola = []
    visitados = []
    padre = {}

    cola.append(inicio)
    visitados.append(inicio)
    padre[inicio] = None

    encontrado = False

    while len(cola) > 0:

        actual = cola.pop(0)

        if actual == destino:
            encontrado = True
            break

        for vecino in grafo[actual]:

            if vecino not in visitados:

                visitados.append(vecino)
                cola.append(vecino)
                padre[vecino] = actual

    if encontrado:

        ruta = []
        actual = destino

        while actual != None:

            ruta.append(actual)
            actual = padre[actual]

        ruta.reverse()

        print("\nCentro encontrado")
        print("Ruta:", " -> ".join(ruta))
        print("Cantidad de conexiones:", len(ruta)-1)

    else:

        print("No existe una ruta hasta el centro solicitado.")



def agregar_nuevo_centro(grafo):

    nuevo = input("\nIngrese el nuevo centro: ").upper()

    if nuevo in grafo:

        print("Ese centro ya existe.")
        return

    grafo[nuevo] = []

    cantidad = int(input("¿Con cuántos centros desea conectarlo?: "))

    for i in range(cantidad):

        conexion = input("Conectar con: ").upper()

        if conexion in grafo:

            grafo[nuevo].append(conexion)
            grafo[conexion].append(nuevo)

        else:

            print("Ese centro no existe.")

    print("\nNueva lista de adyacencia\n")

    for centro in grafo:

        print(centro, "->", " - ".join(grafo[centro]))




def Main():

    grafo = representar_grafo()

    opcion = ""

    while opcion != "7":

        print("\n===== MENÚ =====")
        print("1. Mostrar grafo")
        print("2. Recorrido BFS")
        print("3. Recorrido DFS")
        print("4. Seleccionar punto inicial")
        print("5. Buscar un centro")
        print("6. Agregar un nuevo centro")
        print("7. Salir")

        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            representar_grafo()

        elif opcion == "2":
            inicio = seleccionar_punto_inicial(grafo)
            if inicio != None:
                recorrido_BFS(grafo, inicio)

        elif opcion == "3":
            inicio = seleccionar_punto_inicial(grafo)
            if inicio != None:
                recorrido_DFS(grafo, inicio)

        elif opcion == "4":
            inicio = seleccionar_punto_inicial(grafo)
            if inicio != None:
                recorrido_BFS(grafo, inicio)
                recorrido_DFS(grafo, inicio)

        elif opcion == "5":
            buscar_centro_BFS(grafo)

        elif opcion == "6":
            agregar_nuevo_centro(grafo)

        elif opcion == "7":
            print("Programa finalizado.")

        else:
            print("Opción inválida.")

Main()
```
