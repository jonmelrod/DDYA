### 5) Análisis de complejidad **[15 pts]**

**Justificación:** Sea $V$ el número de estaciones y $E$ el número de conexiones no dirigidas. La lista de adyacencia almacena cada conexión en los dos sentidos, por lo que contiene $2E$ referencias. Tanto BFS como DFS visitan las estaciones alcanzables y revisan sus conexiones; su complejidad general es $\Theta(V+E)$ y el espacio adicional es $O(V)$.

Para expresar el tiempo solo en función de $n=V$, contamos de forma proporcional una operación por estación y una por referencia revisada:

$$T(n)=n+2E$$

- **Mejor caso (red conectada dispersa):** $E=n-1$, entonces $T_{mejor}(n)=n+2(n-1)=3n-2=\Theta(n)$.
- **Peor caso (red completa):** $E=\frac{n(n-1)}{2}$, entonces $T_{peor}(n)=n+n(n-1)=n^2=\Theta(n^2)$.

La gráfica compara ambos casos con escalas verticales independientes para que se aprecien claramente las formas de crecimiento.

## 6) Código funcional **[15 pts]**


```python
from collections import deque

red = {
    "Central": ["Norte", "Museo"],
    "Norte": ["Central", "Parque", "Universidad"],
    "Museo": ["Central", "Hospital"],
    "Parque": ["Norte"],
    "Universidad": ["Norte", "Terminal"],
    "Hospital": ["Museo", "Terminal"],
    "Terminal": ["Universidad", "Hospital"]
}

def buscar_estacion(nombre):
    for estacion in red:
        if estacion.casefold() == nombre.strip().casefold():
            return estacion
    return None

def mostrar_red():
    print("\n--- RED DE ESTACIONES ---")
    for estacion, conexiones in red.items():
        print(f"{estacion}: {', '.join(conexiones)}")


def bfs(inicio):
    visitados = {inicio}
    cola = deque([inicio])
    niveles = {inicio: 0}
    orden = []
    while cola:
        actual = cola.popleft()
        orden.append(actual)
        for vecino in red[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                niveles[vecino] = niveles[actual] + 1
                cola.append(vecino)
    return orden, niveles

def mostrar_bfs(inicio):
    orden, niveles = bfs(inicio)
    print(f"\nRecorrido BFS desde {inicio}:")
    print(" -> ".join(orden))
    print("\nNiveles de las estaciones:")
    for estacion in orden:
        print(f"Nivel {niveles[estacion]}: {estacion}")


def dfs(inicio):
    visitados = set()
    orden = []
    def explorar(actual):
        visitados.add(actual)
        orden.append(actual)
        for vecino in red[actual]:
            if vecino not in visitados:
                explorar(vecino)
    explorar(inicio)
    return orden

def mostrar_dfs(inicio):
    orden = dfs(inicio)
    print(f"\nRecorrido DFS desde {inicio}:")
    print(" -> ".join(orden))


def pedir_estacion(mensaje):
    nombre = input(mensaje).strip()
    estacion = buscar_estacion(nombre)
    if estacion is None:
        print(f"La estación '{nombre}' no existe.")
    return estacion

def ruta_mas_corta(origen, destino):
    cola = deque([origen])
    visitados = {origen}
    anterior = {origen: None}
    while cola:
        actual = cola.popleft()
        if actual == destino:
            break
        for vecino in red[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                anterior[vecino] = actual
                cola.append(vecino)
    if destino not in anterior:
        return None
    ruta = []
    actual = destino
    while actual is not None:
        ruta.append(actual)
        actual = anterior[actual]
    ruta.reverse()
    return ruta

def agregar_estacion():
    nombre = input("\nNombre de la nueva estación: ").strip()
    if not nombre:
        print("Debes escribir un nombre.")
        return
    if buscar_estacion(nombre) is not None:
        print("Esa estación ya está en la red.")
        return
    if len(red) < 2:
        print("Debe haber al menos dos estaciones en la red.")
        return
    red[nombre] = []
    print("\nEstaciones disponibles:")
    print(", ".join(estacion for estacion in red
                     if estacion != nombre))
    while len(red[nombre]) < 2:
        amigo = pedir_estacion("¿Con qué estación quieres conectar?: ")
        if amigo is None:
            continue
        if amigo == nombre:
            print("No puedes conectar una estación consigo misma.")
            continue
        if amigo in red[nombre]:
            print("Ya existe esa conexión.")
            continue
        red[nombre].append(amigo)
        red[amigo].append(nombre)
        print(f"Se conectó {nombre} con {amigo}.")
    print("\nNueva estación agregada.")
    mostrar_red()


def menu():
    while True:
        print("\n======= RED DE TRANSPORTE =======")
        print("1. Mostrar estaciones y conexiones")
        print("2. BFS desde Central")
        print("3. DFS desde Central")
        print("4. BFS y DFS desde otra estación")
        print("5. Buscar la ruta más corta")
        print("6. Agregar una estación")
        print("7. Salir")
        print("=================================")
        opcion = input("Elige una opción: ").strip()
        if opcion == "1":
            mostrar_red()
        elif opcion == "2":
            mostrar_bfs("Central")
        elif opcion == "3":
            mostrar_dfs("Central")
        elif opcion == "4":
            inicio = pedir_estacion(
                "¿Desde qué estación quieres empezar?: "
            )
            if inicio is not None:
                mostrar_bfs(inicio)
                mostrar_dfs(inicio)
        elif opcion == "5":
            origen = pedir_estacion("Estación de origen: ")
            if origen is None:
                continue
            destino = pedir_estacion("Estación de destino: ")
            if destino is None:
                continue
            ruta = ruta_mas_corta(origen, destino)
            if ruta is None:
                print("No hay una ruta entre esas estaciones.")
            else:
                print("\nRuta más corta:")
                print(" -> ".join(ruta))
                print("Número de conexiones:", len(ruta) - 1)
        elif opcion == "6":
            agregar_estacion()
        elif opcion == "7":
            print("Programa finalizado.")
            break
        else:
            print("Opción incorrecta. Elige un número del 1 al 7.")

if __name__ == "__main__":
    menu()
```
