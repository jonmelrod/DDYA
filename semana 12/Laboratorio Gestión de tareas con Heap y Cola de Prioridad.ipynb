### 5) Análisis de complejidad **[15 pts]**

El programa utiliza un Max-Heap para organizar las tareas según su prioridad numérica. La complejidad depende de la cantidad de tareas registradas, representada por n.

Las operaciones de consulta, inserción y extracción aprovechan la estructura de árbol del Heap. Como su altura es aproximadamente log

(n), las operaciones que recorren sus niveles pueden ejecutarse en tiempo logarítmico. La construcción del Heap requiere tiempo lineal, mientras que HeapSort tiene complejidad O(nlogn).

Para el análisis se toma como referencia la operación MAX-HEAPIFY, que compara un elemento con sus hijos y, si es necesario, continúa reorganizando el Heap.

2. Función T(n)

Para MAX-HEAPIFY, podemos expresar el tiempo de ejecución de forma simplificada así:

Mejor caso: T(n)=c, porque el elemento ya está en su posición correcta y no necesita intercambios. Su complejidad es O(1).

Peor caso: T(n)=alog
2
	​

(n)+b, porque el elemento puede bajar por todos los niveles del Heap. Su complejidad es O(logn).

Las constantes a, b y c representan costos de operaciones y no son valores medidos del programa.

## 6) Código funcional **[15 pts]**

```python
# Cada tarea es una lista: [codigo, nombre, prioridad]
# La prioridad esta en la posicion 2 (tarea[2])

def max_heapify(a, n, i):
    """Baja el elemento i hasta su lugar. n = tamano del heap."""
    mayor = i
    izq = 2 * i + 1
    der = 2 * i + 2
    if izq < n and a[izq][2] > a[mayor][2]:
        mayor = izq
    if der < n and a[der][2] > a[mayor][2]:
        mayor = der
    if mayor != i:
        a[i], a[mayor] = a[mayor], a[i]
        max_heapify(a, n, mayor)


def build_max_heap(a):
    """Convierte la lista en un Max-Heap."""
    n = len(a)
    for i in range(n // 2 - 1, -1, -1):
        max_heapify(a, n, i)


def maximum(heap):
    """Devuelve la tarea de la raiz sin eliminarla."""
    if len(heap) == 0:
        return None
    return heap[0]


def extract_max(heap):
    """Elimina y devuelve la tarea de mayor prioridad."""
    if len(heap) == 0:
        return None
    mayor = heap[0]
    heap[0] = heap[-1]
    heap.pop()
    max_heapify(heap, len(heap), 0)
    return mayor


def insertar(heap, tarea):
    """Inserta una tarea y la sube hasta su lugar."""
    heap.append(tarea)
    i = len(heap) - 1
    while i > 0:
        padre = (i - 1) // 2
        if heap[i][2] > heap[padre][2]:
            heap[i], heap[padre] = heap[padre], heap[i]
            i = padre
        else:
            break


def prioridades(lista):
    return [t[2] for t in lista]


def mostrar_arbol(heap):
    """Muestra el heap por niveles."""
    nivel = 0
    inicio = 0
    while inicio < len(heap):
        fin = min(inicio + 2 ** nivel, len(heap))
        print("Nivel", nivel, ":", prioridades(heap[inicio:fin]))
        inicio = fin
        nivel += 1


def heapsort(tareas):
    """Ordena las prioridades de menor a mayor sin sort() ni sorted()."""
    copia = tareas.copy()
    print("Lista original:", prioridades(copia))
    build_max_heap(copia)
    print("Max-Heap:", prioridades(copia))
    for i in range(len(copia) - 1, 0, -1):
        copia[0], copia[i] = copia[i], copia[0]
        max_heapify(copia, i, 0)
    ordenada = prioridades(copia)
    print("Lista ordenada:", ordenada)
    return ordenada


def menu():
    tareas = []   # orden original
    heap = []

    while True:
        print("\n===== MENU =====")
        print("1. Registrar tareas")
        print("2. Construir Max-Heap")
        print("3. Ver tarea de mayor prioridad (MAXIMUM)")
        print("4. Procesar tarea de mayor prioridad (EXTRACT-MAX)")
        print("5. Insertar nueva tarea")
        print("6. HeapSort")
        print("7. Salir")
        opcion = input("Opcion: ")

        if opcion == "1":
            n = int(input("Cantidad de tareas: "))
            tareas = []
            heap = []
            for i in range(n):
                print("Tarea", i + 1)
                codigo = input("  Codigo: ")
                nombre = input("  Nombre: ")
                prio = int(input("  Prioridad: "))
                tareas.append([codigo, nombre, prio])
            print("Lista original de prioridades:", prioridades(tareas))

        elif opcion == "2":
            if len(tareas) == 0:
                print("Primero registre tareas.")
            else:
                heap = tareas.copy()
                build_max_heap(heap)
                print("Lista original:", prioridades(tareas))
                print("Max-Heap:", prioridades(heap))
                mostrar_arbol(heap)

        elif opcion == "3":
            t = maximum(heap)
            if t is None:
                print("No hay tareas en el heap.")
            else:
                print("Tarea de mayor prioridad:", t[0], "-", t[1], "| Prioridad:", t[2])

        elif opcion == "4":
            t = extract_max(heap)
            if t is None:
                print("No hay tareas para procesar.")
            else:
                print("Tarea procesada:", t[0], "-", t[1], "| Prioridad:", t[2])
                print("Nuevo Heap:", prioridades(heap))

        elif opcion == "5":
            codigo = input("Codigo: ")
            nombre = input("Nombre: ")
            prio = int(input("Prioridad: "))
            insertar(heap, [codigo, nombre, prio])
            print("Nuevo Heap:", prioridades(heap))

        elif opcion == "6":
            if len(tareas) == 0:
                print("Primero registre tareas.")
            else:
                heapsort(tareas)

        elif opcion == "7":
            print("Programa finalizado.")
            break

        else:
            print("Opcion no valida.")
```

```python
# Ejecutar primero la celda de funciones. Estas pruebas no piden datos por teclado.
def es_max_heap(a):
    for i in range(len(a)):
        for h in (2 * i + 1, 2 * i + 2):
            if h < len(a) and a[h][2] > a[i][2]:
                return False
    return True


# ---------- Datos del enunciado ----------
tareas = [
    ["T01", "Compilar proyecto", 8],
    ["T02", "Generar backup", 4],
    ["T03", "Renderizar video", 10],
    ["T04", "Procesar correos", 3],
    ["T05", "Ejecutar pruebas", 7],
]

print("=== PUNTO 1: Registro ===")
print("Lista original de prioridades:", prioridades(tareas))
assert prioridades(tareas) == [8, 4, 10, 3, 7]

print("\n=== PUNTO 2: Max-Heap ===")
heap = tareas.copy()
build_max_heap(heap)
print("Max-Heap:", prioridades(heap))
mostrar_arbol(heap)
assert prioridades(heap) == [10, 7, 8, 3, 4]
assert es_max_heap(heap)
assert prioridades(tareas) == [8, 4, 10, 3, 7]   # la original no cambia

print("\n=== PUNTO 3A: MAXIMUM ===")
t = maximum(heap)
print("Tarea de mayor prioridad:", t[0], "-", t[1], "| Prioridad:", t[2])
assert t[0] == "T03" and t[2] == 10
assert len(heap) == 5                              # no elimina nada

print("\n=== PUNTO 3B: EXTRACT-MAX ===")
t = extract_max(heap)
print("Tarea procesada:", t[0], "-", t[1])
print("Nuevo Heap:", prioridades(heap))
assert t[0] == "T03"
assert prioridades(heap) == [8, 7, 4, 3]
assert es_max_heap(heap)

print("\n=== PUNTO 3C: Insertar T06 (prioridad 12) ===")
insertar(heap, ["T06", "Actualizar servidor", 12])
print("Nuevo Heap:", prioridades(heap))
assert heap[0][2] == 12                            # quedo en la raiz
assert es_max_heap(heap)

print("\n=== PUNTO 4: HeapSort ===")
resultado = heapsort(tareas)
assert resultado == [3, 4, 7, 8, 10]

# ---------- PUNTO 5: Prueba final con 6 tareas ----------
print("\n=== PUNTO 5: PRUEBA FINAL (6 tareas) ===")
final = [
    ["T01", "Compilar proyecto", 8],
    ["T02", "Generar backup", 4],
    ["T03", "Renderizar video", 10],
    ["T04", "Procesar correos", 3],
    ["T05", "Ejecutar pruebas", 7],
    ["T06", "Actualizar servidor", 12],
]
print("Registro:", prioridades(final))
hf = final.copy()
build_max_heap(hf)
print("Max-Heap:", prioridades(hf))
assert es_max_heap(hf)
t = maximum(hf)
print("Maximo:", t[0], "-", t[1], "| Prioridad:", t[2])
assert t[2] == 12
t = extract_max(hf)
print("Extraida:", t[0], "-", t[1], "| Nuevo Heap:", prioridades(hf))
assert t[2] == 12 and es_max_heap(hf)
insertar(hf, ["T07", "Analizar logs", 9])
print("Tras insertar T07 (9):", prioridades(hf))
assert es_max_heap(hf) and len(hf) == 6
assert heapsort(final) == [3, 4, 7, 8, 10, 12]

# ---------- Casos limite ----------
print("\n=== CASOS LIMITE ===")
vacio = []
assert maximum(vacio) is None
assert extract_max(vacio) is None
print("Heap vacio: maximum y extract_max devuelven None  OK")

uno = [["T01", "Unica", 5]]
assert extract_max(uno)[2] == 5 and uno == []
print("Heap de un elemento  OK")

repetidos = [["A", "a", 5], ["B", "b", 5], ["C", "c", 5]]
build_max_heap(repetidos)
assert es_max_heap(repetidos)
assert heapsort(repetidos) == [5, 5, 5]
print("Prioridades repetidas  OK")

# Extraer todo debe salir de mayor a menor
h = tareas.copy()
build_max_heap(h)
salida = []
while len(h) > 0:
    salida.append(extract_max(h)[2])
assert salida == [10, 8, 7, 4, 3]
print("Extraer todo sale en orden descendente:", salida)

print("\nTODAS LAS PRUEBAS PASARON")
```
