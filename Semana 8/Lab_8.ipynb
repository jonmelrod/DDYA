```python
import matplotlib.pyplot as plt



class Nodo:

    def __init__(self, valor):
        self.valor = valor
        self.izquierda = None
        self.derecha = None
        self.altura = 1




def insertar_bst(raiz, valor):

    if raiz is None:
        return Nodo(valor)

    if valor < raiz.valor:
        raiz.izquierda = insertar_bst(raiz.izquierda, valor)


    elif valor > raiz.valor:
        raiz.derecha = insertar_bst(raiz.derecha, valor)

    return raiz


def arbol_binario(codigos):

    raiz = None

    for codigo in codigos:
        raiz = insertar_bst(raiz, codigo)

    return raiz



def altura(nodo):

    if nodo is None:
        return 0

    return nodo.altura



def factor_balance(nodo):

    if nodo is None:
        return 0

    return altura(nodo.izquierda) - altura(nodo.derecha)



def rotacion_derecha(y):

    x = y.izquierda
    t2 = x.derecha

    x.derecha = y
    y.izquierda = t2

    y.altura = 1 + max(altura(y.izquierda), altura(y.derecha))
    x.altura = 1 + max(altura(x.izquierda), altura(x.derecha))

    return x


def rotacion_izquierda(x):

    y = x.derecha
    t2 = y.izquierda

    y.izquierda = x
    x.derecha = t2

    x.altura = 1 + max(altura(x.izquierda), altura(x.derecha))
    y.altura = 1 + max(altura(y.izquierda), altura(y.derecha))

    return y


def insertar_avl(raiz, valor):

    if raiz is None:
        return Nodo(valor)

    if valor < raiz.valor:
        raiz.izquierda = insertar_avl(raiz.izquierda, valor)

    elif valor > raiz.valor:
        raiz.derecha = insertar_avl(raiz.derecha, valor)

    else:
        return raiz


    raiz.altura = 1 + max(altura(raiz.izquierda), altura(raiz.derecha))


    balance = factor_balance(raiz)

    if balance > 1 and valor < raiz.izquierda.valor:
        return rotacion_derecha(raiz)


    if balance < -1 and valor > raiz.derecha.valor:
        return rotacion_izquierda(raiz)


    if balance > 1 and valor > raiz.izquierda.valor:
        raiz.izquierda = rotacion_izquierda(raiz.izquierda)
        return rotacion_derecha(raiz)


    if balance < -1 and valor < raiz.derecha.valor:
        raiz.derecha = rotacion_derecha(raiz.derecha)
        return rotacion_izquierda(raiz)

    return raiz


# Construir el árbol AVL
def arbol_avl(codigos):

    raiz = None

    for codigo in codigos:
        raiz = insertar_avl(raiz, codigo)

    return raiz


def recorrido_preorden(raiz):

    if raiz is not None:

        print(raiz.valor, end=" ")

        recorrido_preorden(raiz.izquierda)

        recorrido_preorden(raiz.derecha)


def recorrido_inorden(raiz):

    if raiz is not None:

        recorrido_inorden(raiz.izquierda)

        print(raiz.valor, end=" ")

        recorrido_inorden(raiz.derecha)



def recorrido_postorden(raiz):

    if raiz is not None:

        recorrido_postorden(raiz.izquierda)

        recorrido_postorden(raiz.derecha)

        print(raiz.valor, end=" ")




def obtener_posiciones(nodo, x, y, espacio, posiciones, conexiones):

    if nodo is None:
        return

    posiciones[nodo] = (x, y)

    if nodo.izquierda is not None:

        conexiones.append((nodo, nodo.izquierda))

        obtener_posiciones(
            nodo.izquierda,
            x - espacio,
            y - 1,
            espacio / 2,
            posiciones,
            conexiones
        )

    if nodo.derecha is not None:

        conexiones.append((nodo, nodo.derecha))

        obtener_posiciones(
            nodo.derecha,
            x + espacio,
            y - 1,
            espacio / 2,
            posiciones,
            conexiones
        )


def visualizar(raiz, titulo):

    posiciones = {}
    conexiones = []

    obtener_posiciones(raiz, 0, 0, 8, posiciones, conexiones)

    plt.figure(figsize=(10,6))
    plt.title(titulo)


    for padre, hijo in conexiones:

        x1, y1 = posiciones[padre]
        x2, y2 = posiciones[hijo]

        plt.plot([x1, x2], [y1, y2])


    for nodo, (x, y) in posiciones.items():

        plt.scatter(x, y, s=900)

        plt.text(
            x,
            y,
            str(nodo.valor),
            ha="center",
            va="center",
            color="white",
            fontsize=10
        )

    plt.axis("off")
    plt.show()



def Main():

    cantidad = int(input("Ingrese la cantidad de productos: "))

    codigos = []

    for i in range(cantidad):

        codigo = int(input(f"Ingrese el código {i+1}: "))
        codigos.append(codigo)

    print("\nLista de códigos:")
    print(codigos)


    bst = arbol_binario(codigos)


    avl = arbol_avl(codigos)

    print("\n========== BST ==========")

    print("Preorden:")
    recorrido_preorden(bst)

    print("\n")

    print("Inorden:")
    recorrido_inorden(bst)

    print("\n")

    print("Postorden:")
    recorrido_postorden(bst)

    print("\n")

    print("========== AVL ==========")

    print("Preorden:")
    recorrido_preorden(avl)

    print("\n")

    print("Inorden:")
    recorrido_inorden(avl)

    print("\n")

    print("Postorden:")
    recorrido_postorden(avl)

    print("\n")

    visualizar(bst, "Árbol Binario de Búsqueda")

    visualizar(avl, "Árbol AVL")


Main()
```
