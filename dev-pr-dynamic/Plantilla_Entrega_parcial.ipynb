-En el mes 0 se registraron 0 nuevos usuarios.
-En el mes 1 se registro 1 nuevo usuario.
-A partir del mes 2, el numero de nuevos usuarios es igual a la suma de los nuevos usuarios dos meses anteriores, debido al efecto combinado de referidos del mes anterior y recomendaciones acumuladas del mes previo.

Relacion de recurrencia:

T(n) = 0 si n = 0
T(n) = 1 si n = 1
T(n) = T(n - 1) + T(n - 2) si n >= 2

## 6) Código funcional **[15 pts]**

```python
# Escribe tu código aquí.
def memoizacion(n):
    memo = {}
    if n == 0:
        return 0
    if n == 1:
        return 1
    if n in memo:
        return memo[n]
    memo[n] = memoizacion(n - 1) + memoizacion(n - 2)
    return memo[n]
def main():
    n = int(input("Ingrese el valor de n: "))
    resultado = memoizacion(n)
    print(f"El resultado de memoizacion({n}) es: {resultado}")

main()
```
