## 1) Descripción del problema **[1 pts]**

Sistema de atención de una tienda

Una tienda desea desarrollar un pequeño sistema para controlar la atención de clientes, registrar las operaciones realizadas y administrar su catálogo de productos.

Para resolver el problema se deberán utilizar tres estructuras de datos diferentes:

Cola: para administrar los clientes que esperan ser atendidos.
Pila: para guardar el historial de las últimas operaciones realizadas.
Lista doblemente enlazada: para administrar y recorrer el catálogo de productos.
Punto 1. Cola de clientes
Los clientes son atendidos en el mismo orden en que llegan a la tienda.

El usuario deberá ingresar inicialmente una cantidad n de clientes.

Ejemplo:

¿Cuántos clientes desea registrar?: 4 Ingrese el nombre del cliente 1: Ana Ingrese el nombre del cliente 2: Carlos Ingrese el nombre del cliente 3: Laura Ingrese el nombre del cliente 4: Mateo
La cola inicialmente sería:

Ana -> Carlos -> Laura -> Mateo
El programa deberá permitir:

Agregar un nuevo cliente a la cola.
Atender al siguiente cliente.
Mostrar los clientes que siguen esperando.
Por ejemplo:

Atendiendo a: Ana Clientes esperando: Carlos -> Laura -> Mateo
Cuando llegue un nuevo cliente:

Ingrese el nuevo cliente: Sofía
La cola deberá quedar:

Carlos -> Laura -> Mateo -> Sofía
Punto 2. Pila de operaciones
Cada vez que se atienda un cliente, su nombre deberá guardarse en una pila de clientes atendidos.

Por ejemplo, después de atender:

Ana Carlos Laura
La pila sería conceptualmente:

| Laura  | <- Tope | Carlos | | Ana    | ---------
El programa deberá permitir:

Agregar a la pila cada cliente que sea atendido.
Mostrar el último cliente atendido.
Deshacer la última atención.
Si se selecciona la opción:

Deshacer última atención
y Laura fue la última cliente atendida, deberá retirarse de la pila.

La pila quedaría:

| Carlos | <- Tope | Ana    | ---------
Además, Laura deberá regresar al inicio de la cola de clientes pendientes.

Punto 3. Lista doblemente enlazada de productos
La tienda también necesita mantener un catálogo de productos.

Cada nodo deberá almacenar:

Código Nombre Precio
Por ejemplo:

101 - Arroz - $4.500 102 - Leche - $3.800 103 - Pan - $2.500 104 - Café - $12.000
Estos productos deberán almacenarse utilizando una lista doblemente enlazada.

Cada nodo tendrá:

anterior <- Producto -> siguiente
Ejemplo:

NULL <- Arroz <-> Leche <-> Pan <-> Café -> NULL
El programa deberá permitir:

Agregar un producto al final de la lista.
Mostrar todos los productos desde el primero hasta el último.
Mostrar los productos desde el último hasta el primero.
Buscar un producto utilizando su código.
Eliminar un producto utilizando su código.
Por ejemplo:

Ingrese el código que desea eliminar: 103
Después de eliminar el producto Pan, la lista deberá quedar:

NULL <- Arroz <-> Leche <-> Café -> NULL

````python
cola_clientes = []
pila_atendidos = []
productos = []


def cola():

    opcion = ""

    while opcion != "4":

        print("\n------ COLA DE CLIENTES ------")
        print("1. Agregar cliente")
        print("2. Atender cliente")
        print("3. Mostrar clientes")
        print("4. Volver")

        opcion = input("Seleccione una opcion: ")

        if opcion == "1":

            nombre = input("Ingrese el nombre del cliente: ")
            cola_clientes.append(nombre)

            print("Cliente agregado.")

        elif opcion == "2":

            if len(cola_clientes) == 0:
                print("No hay clientes en espera.")

            else:

                atendido = cola_clientes.pop(0)

                print("Cliente atendido:", atendido)

                pila_atendidos.append(atendido)

        elif opcion == "3":

            if len(cola_clientes) == 0:
                print("No hay clientes esperando.")

            else:

                print("Clientes esperando:")

                for cliente in cola_clientes:
                    print(cliente)

        elif opcion != "4":

            print("Opcion incorrecta.")


def pila():

    opcion = ""

    while opcion != "4":

        print("\n------ PILA DE ATENDIDOS ------")
        print("1. Mostrar ultimo atendido")
        print("2. Deshacer ultima atencion")
        print("3. Mostrar historial")
        print("4. Volver")

        opcion = input("Seleccione una opcion: ")

        if opcion == "1":

            if len(pila_atendidos) == 0:

                print("No hay clientes atendidos.")

            else:

                print("Ultimo atendido:", pila_atendidos[-1])

        elif opcion == "2":

            if len(pila_atendidos) == 0:

                print("No hay clientes para deshacer.")

            else:

                cliente = pila_atendidos.pop()

                cola_clientes.insert(0, cliente)

                print(cliente, "regreso al inicio de la cola.")

        elif opcion == "3":

            if len(pila_atendidos) == 0:

                print("La pila esta vacia.")

            else:

                print("Historial de clientes atendidos:")

                for i in range(len(pila_atendidos)-1, -1, -1):

                    print(pila_atendidos[i])

        elif opcion != "4":

            print("Opcion incorrecta.")



def lista_doble_enlazada():

    opcion = ""

    while opcion != "6":

        print("\n------ CATALOGO DE PRODUCTOS ------")
        print("1. Agregar producto")
        print("2. Mostrar productos")
        print("3. Mostrar productos al reves")
        print("4. Buscar producto")
        print("5. Eliminar producto")
        print("6. Volver")

        opcion = input("Seleccione una opcion: ")

        if opcion == "1":

            codigo = input("Codigo: ")
            nombre = input("Nombre: ")
            precio = input("Precio: ")

            producto = {
                "codigo": codigo,
                "nombre": nombre,
                "precio": precio
            }

            productos.append(producto)

            print("Producto agregado.")

        elif opcion == "2":

            if len(productos) == 0:

                print("No hay productos.")

            else:

                print("\nCatalogo de productos:\n")

                for producto in productos:

                    print(producto["codigo"], "-", producto["nombre"], "-", producto["precio"])

        elif opcion == "3":

            if len(productos) == 0:

                print("No hay productos.")

            else:

                print("\nCatalogo en orden inverso:\n")

                for i in range(len(productos)-1, -1, -1):

                    producto = productos[i]

                    print(producto["codigo"], "-", producto["nombre"], "-", producto["precio"])

        elif opcion == "4":

            codigo = input("Ingrese el codigo: ")

            encontrado = False

            for producto in productos:

                if producto["codigo"] == codigo:

                    print(producto["codigo"], "-", producto["nombre"], "-", producto["precio"])

                    encontrado = True

            if encontrado == False:

                print("Producto no encontrado.")

        elif opcion == "5":

            codigo = input("Ingrese el codigo a eliminar: ")

            eliminado = False

            for producto in productos:

                if producto["codigo"] == codigo:

                    productos.remove(producto)

                    eliminado = True

                    print("Producto eliminado.")

                    break

            if eliminado == False:

                print("Producto no encontrado.")

        elif opcion != "6":

            print("Opcion incorrecta.")




def main():

    cantidad = int(input("¿Cuantos clientes desea registrar?: "))

    for i in range(cantidad):

        nombre = input("Ingrese el nombre del cliente: ")

        cola_clientes.append(nombre)

    opcion = ""

    while opcion != "4":

        print("\n========== MENU PRINCIPAL ==========")
        print("1. Cola de clientes")
        print("2. Pila de atendidos")
        print("3. Catalogo de productos")
        print("4. Salir")

        opcion = input("Seleccione una opcion: ")

        if opcion == "1":

            cola()

        elif opcion == "2":

            pila()

        elif opcion == "3":

            lista_doble_enlazada()

        elif opcion != "4":

            print("Opcion incorrecta.")

    print("\nPrograma finalizado.")


main()
````

## Pruebas de enunciado

Se validan los escenarios indicados en la descripción del problema:

- Cola FIFO: el primer cliente que llega debe ser atendido primero.
- Pila LIFO: el último cliente atendido debe ser el primero en deshacerse.
- Catálogo de productos: se debe poder buscar, mostrar y eliminar por código.

### Caso 1: atención de clientes
Entrada usada: ["Ana", "Carlos", "Laura", "Mateo"]
Salida esperada: atender a "Ana" y dejar la cola como ["Carlos", "Laura", "Mateo"]

### Caso 2: historial de clientes atendidos
Entrada usada: ["Ana", "Carlos", "Laura"]
Salida esperada: el último atendido es "Laura" y al deshacer la última atención queda ["Ana", "Carlos"]

### Caso 3: eliminación de un producto
Entrada usada: catálogo con códigos [101, 102, 103, 104]
Salida esperada: tras eliminar 103, la lista queda [101, 102, 104]


````python
input, output = ["Ana", "Carlos", "Laura", "Mateo"], ["Carlos", "Laura", "Mateo"]
assert input.pop(0) == "Ana"
assert input == output
print(f"Paso caso de prueba {input} -> {output}")

# Caso de pila
pila = ["Ana", "Carlos", "Laura"]
resultado = pila.pop()
assert resultado == "Laura"
assert pila == ["Ana", "Carlos"]
print(f"Paso caso de prueba pila {pila} -> {'Laura'}")

# Caso de catálogo
catalogo = [101, 102, 103, 104]
catalogo = [cod for cod in catalogo if cod != 103]
assert catalogo == [101, 102, 104]
print(f"Paso caso de prueba catalogo {catalogo} -> {[101, 102, 104]}")
````

#### Código base para benchmarking

````python
# Calcula el tiempo de ejecucion con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecucion
# TENER EN CUENTA ESTAS ENTRADAS PUEDEN VARIAR SEGÚN EL PROBLEMA


def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    from random import randint
    import time
    results: list[tuple[int, float]] = []


    # AJUSTAR LAS ENTRADAS SEGÚN EL PROBLEMA
    for size in range(1, n, step):
        input = size

        ti = time.perf_counter()
        output = input # Ajustar a su función
        tf = time.perf_counter()


        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
    return results


# === CODIGO PARA DIBUJAR ===


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y)
    plt.show()


# === LLAMADO ===
tiempos = bench(10, step=1)


x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]


dibujar(x, y)
````
